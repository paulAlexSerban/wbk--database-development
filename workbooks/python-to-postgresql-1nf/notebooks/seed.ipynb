## What is 1NF?

A relation is in **First Normal Form (1NF)** when:

1. Every column holds **atomic** (indivisible) values — no lists, no delimited strings.
2. There are **no repeating groups** — each attribute appears once per row.
3. Each row/column intersection holds **exactly one value**.

### Before (violates 1NF)

`orders_unnormalized` stores multiple phone numbers and line items as delimited strings:

| customer_name | customer_phone_numbers | line_items |
|---|---|---|
| Ada Lovelace | 555-0101,555-0199 | SKU1:2:1999,SKU2:1:499 |

### After (1NF)

Atomic values across related tables: `customers`, `customer_phone_numbers`, `orders`, `order_items`.
