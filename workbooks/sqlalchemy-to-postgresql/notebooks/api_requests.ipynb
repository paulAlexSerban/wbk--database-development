# Flask API requests

Use this notebook instead of Postman to explore the Flask service.

Run `make compose_up` and `make seed`, then open this notebook at `http://localhost:8888`.

```python
import json
import os

import requests

BASE_URL = os.environ.get("API_BASE_URL", "http://web:5000")


def api(method, path, **kwargs):
    response = requests.request(method, f"{BASE_URL}{path}", **kwargs)
    print(response.status_code)
    try:
        print(json.dumps(response.json(), indent=2))
    except ValueError:
        print(response.text)
    return response
```

## Health

```python
api("GET", "/health")
```

## Customers

```python
api("GET", "/customers")
```

```python
api(
    "POST",
    "/customers",
    json={"email": "dana@example.com"},
)
```

## Products (JSONB attributes)

```python
api("GET", "/products")
```

```python
api("GET", "/products", params={"color": "red"})
```

```python
api(
    "POST",
    "/products",
    json={
        "name": "Cap",
        "price_cents": 1500,
        "attributes": {
            "color": "black",
            "size": "one",
            "tags": ["accessory"],
        },
    },
)
```

## Orders (JSONB items)

```python
api("GET", "/orders")
```

```python
api(
    "POST",
    "/orders",
    json={
        "customer_id": 1,
        "status": "pending",
        "items": [
            {
                "product_id": 1,
                "quantity": 2,
                "unit_price_cents": 1999,
            },
            {
                "product_id": 3,
                "quantity": 1,
                "unit_price_cents": 12999,
            },
        ],
    },
)
```
