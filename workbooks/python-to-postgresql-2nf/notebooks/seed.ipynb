## What is 2NF?

A relation is in **Second Normal Form (2NF)** when:

1. It is already in **1NF** (atomic values, no repeating groups).
2. Every non-key attribute depends on the **whole** candidate key — no **partial dependencies**.

### Before (1NF, violates 2NF)

`order_items_1nf` uses composite PK `(order_id, product_sku)`, but `product_name` and
`unit_price_cents` depend only on `product_sku`:

| order_id | product_sku | product_name | unit_price_cents | quantity |
|---|---|---|---|---|
| 1 | ABC123 | Widget | 1999 | 2 |
| 2 | ABC123 | Widget | 1999 | 1 |

Product attributes are repeated on every line that uses the same SKU.

### After (2NF)

- `products(sku, name, unit_price_cents)` — attributes that depend on SKU alone
- `order_items(order_id, product_sku, quantity)` — quantity depends on the full key
