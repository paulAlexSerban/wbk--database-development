## What is 3NF?

A relation is in **Third Normal Form (3NF)** when:

1. It is already in **2NF**.
2. There are **no transitive dependencies** — non-key attributes must depend only on the key,
   not on other non-key attributes.

### Before (2NF, violates 3NF)

`products_2nf` uses `sku` as PK, but `supplier_name` / `supplier_city` depend on
`supplier_id` (`sku → supplier_id → supplier_*`):

| sku | name | supplier_id | supplier_name | supplier_city |
|---|---|---|---|---|
| ABC123 | Widget | 1 | Acme Corp | Springfield |
| DEF456 | Gadget | 1 | Acme Corp | Springfield |

Supplier attributes are repeated on every product from the same supplier.

### After (3NF)

- `suppliers(id, name, city)` — attributes that depend on supplier_id
- `products(sku, name, unit_price_cents, supplier_id)` — FK only, no transitive attrs
